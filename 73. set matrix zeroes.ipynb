```python
matrix = [[0,1,2,0],[3,4,5,2],[1,3,1,5]]
n = len(matrix)
m = len(matrix[0])
col0 = 1
n, m
```

```python
for i in range(n-1, -1, -1):
    for j in range(m-1, -1, -1):
        print(i, j)
        print(matrix[i][j])
```

```python
for i in range(n):
    col0 = 0 if matrix[i][0] == 0 else col0
    for j in range(1, m):
        if matrix[i][j] == 0:
            matrix[i][0] = matrix[0][j] = 0

for i in range(n-1, -1, -1):
    for j in range(m-1, 0, -1):
        print(i, j)
        if matrix[0][j] == 0 or matrix[i][0] == 0:
            matrix[i][j] = 0
    matrix[i][0] = 0 if col0 == 0 else matrix[i][0]
        
```
