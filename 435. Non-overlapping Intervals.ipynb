```python
intervals = [[0,2],[1,3],[2,4],[3,5],[4,6]]
```

```python
if len(intervals) == 0:
    print("0")
```

```python
intervals.sort(key=lambda x:x[1])
```

```python
intervals
```

```python
count = 1

prev = 0

for i in range(1, len(intervals)):
    if intervals[i][0] >= intervals[prev][1]:
        prev = i
        count = count + 1
```

```python
len(intervals) - count
```
