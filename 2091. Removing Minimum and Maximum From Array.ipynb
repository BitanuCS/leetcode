```python
nums = [2,10,7,5,4,1,8,6]
n = len(nums)
```

```python
mini = min(nums[0], nums[1])
maxi = max(nums[0], nums[1])
mini_i, maxi_i = 0, 1

if len(nums) < 2:
    print("1")
if len(nums) < 3:
    print("2")

for i in range(2, len(nums)):
    if mini > nums[i]:
        mini = nums[i]
        mini_i = i 
    if maxi < nums[i]:
        maxi = nums[i]
        maxi_i = i

print(mini, maxi, mini_i, maxi_i)

mid = len(nums)//2

if mini_i <= mid and maxi_i <= mid:
    mini = max(mini_i, maxi_i)
    print(mini + 1)
elif mini_i >= mid and maxi_i >= mid:
    maxi = min(mini_i, maxi_i)
    print(n-maxi +1)
elif mini_i <= mid and maxi_i >= mid:
    print(mini + 1 + (n-maxi)+1)
elif mini_i >= mid and maxi_i <= mid:
    print(maxi + 1 + (n-mini)+1)
# print((min_i + max_i))
```
