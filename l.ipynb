```python
nums = [-1,0,3,5,7,8,9,12]
target = 9
```

```python
def bin_search(nums, l, r):
    # ret = 0
    mid = int((l+r)/2)   
    # print("mid", mid)
    if(nums[mid] < target):
        return bin_search(nums, mid, r)
    if(nums[mid] > target):
        return bin_search(nums, l, mid)
    if(nums[mid] == target):
        return mid
    # return ret
```

```python
l = 0
r = len(nums) -1
res = bin_search(nums,l,r)    
print("index", res)
```

```python
score= [5,4,3,2,1]
```

```python
score1 = [-s for s in score]
heapq.heapify(score1)
if (len(score1)>0):
    f = abs(heapq.heappop(score1))
    f1 = abs(heapq.heappop(score1))
    f2 = abs(heapq.heappop(score1))
for i in range(len(score)):
    if (len(score1)>0):
        if (f == score[i]):
            score[i] = "Gold"
        if (f1 == score[i]):
            score[i] = "Silver"
        if (f2 == score[i]):
            score[i] = "Bronze"
r = 4
for i in range(1, len(score)):
    x = abs(heapq.heappop(score1))
    if (x == score[i]):
        score[i] = str(r) + "th"
        r+=1

x = abs(heapq.heappop(score1))
for i in range(1, len(score)):
    if (x == score[i]):
        score[i] = str(r) + "th"
```
