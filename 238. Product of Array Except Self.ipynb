```python
nums = [1,2,3,4]
answer = []
```

```python
left = [1] * len(nums)
right = [1] * len(nums)

# left[0] = nums[0]
# right[len(nums)-1] = nums[len(nums)-1]

left, right
```

```python
for i in range(1, len(nums)):
    left[i] = left[i-1] * nums[i-1]
```

```python
left
```

```python
for i in range(len(nums)-2, -1, -1):
    right[i] = right[i+1] * nums[i+1]
```

```python
right
```

```python
for i in range(len(nums)):
    answer[i] = left[i] * right[i]
```
