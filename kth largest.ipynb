```python
import random
nums = [3,2,1,5,6,4]
k = 2
```

```python
def findindex(l, h, pivot_index, nums):
    nums[l], nums[pivot_index] = nums[pivot_index], nums[l]
    pivot_value = nums[l]
    ind = l + 1
    for i in range(l+1, h+1):
        if nums[i] > pivot_value:
            nums[ind], nums[i] = nums[i], nums[ind]
            ind += 1
    
    nums[ind-1], nums[l] = nums[l], nums[ind-1]
    return ind - 1
            
```

```python
l, h = 0, len(nums) - 1

while True:
    pivot_index = random.randint(l, h)
    pivot_index = findindex(l, h, pivot_index, nums)
    if pivot_index == k - 1:
        print(nums[pivot_index])
        break
    elif pivot_index > k - 1:
        h = pivot_index - 1
    else:
        l = pivot_index + 1

```
