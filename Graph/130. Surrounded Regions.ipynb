```python
board = [["X","X","X","X"],["X","O","O","X"],["X","X","O","X"],["X","O","X","X"]]
```

```python
def dfs(r, c):
    if r < 0 or c < 0 or r >= rows or c >= cols or board[r][c] != 'O':
        return
    
    board[r][c] = 'T'
    dfs(r-1, c)
    dfs(r+1, c)
    dfs(r, c-1)
    dfs(r, c+1)


rows = len(board)
cols = len(board[0])

for r in range(rows):
    for c in range(cols):
        if board[r][c] == 'O' and (r == 0 or c == 0 or r == rows-1 or c == cols-1):
            dfs(r,c)

for r in range(rows):
    for c in range(cols):
        if board[r][c] == 'O':
            board[r][c] = 'X'

for r in range(rows):
    for c in range(cols):
        if board[r][c] == 'T':
            board[r][c] = 'O'

print(board)
```
