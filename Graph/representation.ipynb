```python
def creategraph(v, edges):
    adj = [[] for _ in range(v)]
    
    for it in edges:
        u = it[0]
        v = it[1]
        
        adj[u].append(v)
        
        adj[v].append(u)
    
    return adj
        
```

```python

v = 10

edges = [[1, 2], [2,3], [2,4], [4,5], [5,8], [7, 8], [6,7], [6,9], [1, 6]]

adj_list = creategraph(v, edges)
```

```python
print("Adjacency List Representation:")
for i in range(v):
        
    # Print the vertex
    print(f"{i}:", end=" ")
    for j in adj_list[i]:
            
        # Print its adjacent
        print(j, end=" ")
    print()
```

```python
from collections import deque


def bfs(src):
    
    res = []
    visited = [False] * len(adj_list)
    
    # src = 0

    q = deque([src])
    visited[src] = True      # mark when you ENQUEUE, not when you dequeue

    while q:
        node = q.popleft()   # take from the FRONT - this is what makes it BFS
        res.append(node)

        for nei in adj_list[node]:
            if not visited[nei]:
                visited[nei] = True
                q.append(nei)

    return res

print(bfs(1))
```
