```python
from collections import deque

from graph_utils import creategraph, print_graph

def bfs(src):
    res = []
    
    visited = [False] * len(adj_list)
    
    q = deque([src])
    visited[src] = True
    
    while q:
        node = q.popleft()
        res.append(node)

        for n in adj_list[node]:
            if not visited[n]:
                visited[n] = True
                q.append(n)
    
    return res

bfs(1)



edges = [[1, 2], [2, 3], [2, 4], [4, 5], [5, 8], [7, 8], [6, 7], [6, 9], [1, 6]]

adj_list = creategraph(9, edges)
print_graph(adj_list)

print(bfs(1))
```
