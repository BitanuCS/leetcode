```python
V = 4
E = 3
edges = [[3, 0], [1, 0], [2, 0]]
```

```python
result = []
indegrees = [0 for _ in range(V)]
adj_list = [[] for _ in range(V)]
```

```python
from collections import deque
for u, v in edges:
    adj_list[u].append(v)
    indegrees[v] += 1

q = deque()

for i in range(V):
    if indegrees[i] == 0:
        q.append(i)

while q:
    node = q.popleft()
    result.append(node)
    for adj_node in adj_list[node]:
        indegrees[adj_node] -= 1
        if indegrees[adj_node] == 0:
            q.append(adj_node)

print(result)
```
