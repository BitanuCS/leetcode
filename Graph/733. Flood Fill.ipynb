```python
image = [[1,1,1],[1,1,0],[1,0,1]]
sr = 1
sc = 1
color = 2
```

```python
from copy import deepcopy
from collections import deque

if image[sr][sc] == color:
    print(image)

m = len(image)
n = len(image[0])
visited = deepcopy(image)
temp = visited[sr][sc]
q = deque()
q.append((sr,sc))

while q:
    x, y = q.popleft()
    visited[x][y] = color
    for xx, yy in [(-1,0), (1,0), (0,-1), (0,1)]:
        
        new_sr = x+xx
        new_sc = y+yy
        
        if new_sc<0 or new_sr<0 or new_sc>=n or new_sr>=m:
            continue
        if visited[new_sr][new_sc] != temp:
            continue
            
        q.append((new_sr,new_sc))
print(visited)
```
