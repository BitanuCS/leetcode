```python
from collections import deque
isConnected = [[1,1,0],[1,1,0],[0,0,1]]
```

```python
n = len(isConnected)
visited = [False] * n


def bfs(start):
    q = deque([start])
    visited[start] = True
    
    while q:
        node = q.popleft()
        
        for nei in range(n):
            if isConnected[node][nei] == 1 and not visited[nei]:
                visited[nei] = True
                q.append(nei)
                
count = 0
for i in range(n):
    if visited[i] == 0:
        count += 1
        bfs(i)
    
print(count)
```
