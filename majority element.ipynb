```python
nums = [7, 0, 0, 1, 3, 3,3,3,3, 2, 7, 7]
```

```python
n = len(nums)
```

```python
hashmap = {}
```

```python
for num in nums:
    hashmap[num] = hashmap.get(num, 0) + 1
        
```

```python
hashmap
```

```python
max(hashmap, key=hashmap.get)
```

```python
data = {7: 3, 0: 2, 1: 1, 3: 10, 2: 1}

# This says: "Give me the key where the value (data.get) is the highest"
max_key = max(data, key=data.get)

print(max_key)
```
