```python
fruits = [1,2,3,2,2]
```

```python
max_length = 0
my_dict = {}
l = r = 0

while r < len(fruits):
    my_dict[fruits[r]] = my_dict.get(fruits[r], 0) + 1
    
    if len(my_dict) > 2:
        my_dict[fruits[l]] -= 1
        if my_dict[fruits[l]] == 0:
            del my_dict[fruits[l]]
        l += 1
    
    if len(my_dict) <= 2:
        max_length = max(max_length, r-l+1)

    r+=1
print(max_length)
```
