```python
cardPoints = [9,7,7,9,7,7,9]
k = 7
```

```python
n = len(cardPoints)
```

```python
maxi = 0
ls = rs = 0

if n == k:
    print(sum(cardPoints))

for i in range(k):
    ls += cardPoints[i]
    
maxi = ls
rp = n-1

for i in range(k-1,-1,-1):
    ls -= cardPoints[i]
    rs += cardPoints[rp]
    
    maxi = max(maxi, ls+rs)
    rp -= 1

print(maxi)
```
