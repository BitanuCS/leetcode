```python
nums = [3,1,-2,-5,2,-4]
```

```python
ans = [0] * len(nums)
p = 0
n = 0

for i in range(0, len(nums)):
    if nums[i] >= 0:
        # print(nums[i])
        
        a = (2 * p)
        print("a",a)
        ans[a] = nums[i]
        p += 1
    if nums[i] < 0:
        # print(nums[i])
        b = (2 * n)+1
        print("b",b)
        ans[b] = nums[i]
        n += 1
```
