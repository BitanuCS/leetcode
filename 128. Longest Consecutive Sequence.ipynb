```python
# LeetCode submission form - O(n) time, O(n) space
class Solution:
    def longestConsecutive(self, nums: list[int]) -> int:
        num_set = set(nums)
        longest = 0

        for num in num_set:              # the SET, never the list
            if num - 1 not in num_set:   # only start at the head of a run
                current = 1
                while num + current in num_set:
                    current += 1
                longest = max(longest, current)

        return longest


print(Solution().longestConsecutive([100, 4, 200, 1, 3, 2]))   # 4
print(Solution().longestConsecutive([0,3,7,2,5,8,4,6,0,1]))    # 9
print(Solution().longestConsecutive([]))                       # 0
```
